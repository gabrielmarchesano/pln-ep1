```python
import pandas as pd

# para tabelas c/separador em português 

mydec = "," 
arq = "train.xlsx" 
df = pd.read_excel(arq, na_values=['na'], decimal=mydec) 
print(df.shape) 
print(df.columns)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer 
from sklearn.linear_model import LogisticRegression 
from sklearn.metrics import f1_score 

vect = TfidfVectorizer () 
X = vect.fit_transform(df['resp_text'])       # a couna de texto a vetorizar 
Y = df['clarity']

clf = LogisticRegression(class_weight='balanced').fit(X, Y) 
predicted = clf.predict(X)
score = f1_score(Y ,predicted,average='macro') 

print("\n\nTF-IDF =====>{:2.2f}\n\n".format(score).replace(".",","))
```
